# `if` Statement

```python
'''
if condition:
    # code block executed if condition is True
'''

x = 10
if x > 0:
    print("x is positive")
```

# `if-else` Statement

```python
'''
if condition:
    # if block
else:
    # else block
'''

x = -5
if x >= 0:
    print("Non-negative")
else:
    print("Negative")
```

# `if-elif-else` Ladder

```python
'''
if condition1:
    # block1
elif condition2:
    # block2
elif condition3:
    # block3
else:
    # default block
'''

score = 85

if score >= 90:
    print("Grade A")
elif score >= 75:
    print("Grade B")
elif score >= 60:
    print("Grade C")
else:
    print("Fail")
```

# Nested Conditional Statements

```python
x = 10
if x > 0:
    if x % 2 == 0:
        print("Positive even number")
    else:
        print("Positive odd number")
else:
    print("Not a positive number")

# Best Practice: Avoid deep nesting; use logical operators instead for clarity:
if x > 0 and x % 2 == 0:
    print("Positive even number")
```

# Ternary (Conditional) Expression

```python
# value_if_true if condition else value_if_false

age = 17
status = "Adult" if age >= 18 else "Minor"
print(status)
```

# `match-case`

```python
'''
match variable:
    case value1:
        # block1
    case value2:
        # block2
    case _:
        # default block
'''

command = "start"

match command:
    case "start":
        print("System starting...")
    case "stop":
        print("System stopping...")
    case _:
        print("Unknown command")
```

# `else` with Loops
An `else` block can be used with `for` or `while`. It runs only if the loop completes normally (no `break`).

```python
for i in range(3):
    print(i)
else:
    print("Loop finished")

# Break example
for i in range(5):
    if i == 3:
        break
    print(i)
else:
    print("This won't print")
```

```python
'''
match variable_name:
   case 'pattern 1' : statement 1
   case 'pattern 2' : statement 2
   ...
   case 'pattern n' : statement n
'''

def weekday(n):
   match n:
      case 0: return "Monday"
      case 1: return "Tuesday"
      case 2: return "Wednesday"
      case 3: return "Thursday"
      case 4: return "Friday"
      case 5: return "Saturday"
      case 6: return "Sunday"
      case _: return "Invalid day number"
print(weekday(3))
print(weekday(6))
print(weekday(7))
```

```python
def greeting(details):
   match details:
      case [time, name]:
         return f'Good {time} {name}!'
      case [time, *names]:
         msg = ''
         for name in names:
            msg += f'Good {time} {name}!\n'
         return msg

print(greeting(["Morning", "Ravi"]))
print(greeting(["Afternoon","Guest"]))
print(greeting(["Evening", "Kajal", "Praveen", "Lata"]))
```

```python
def intr(details):
   match details:
      case [amt, duration] if amt < 10000:
         return amt * 10 * duration / 100
      case [amt, duration] if amt >= 10000:
         return amt * 15 * duration / 100
print("Interest = ", intr([5000,5]))
print("Interest = ", intr([15000,3]))
```
