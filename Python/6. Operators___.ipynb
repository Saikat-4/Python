| **7**      | `+`, `-`                                                                      | **Addition and subtraction**                                 | Left-to-right            |
| **8**      | `<<`, `>>`                                                                    | **Bitwise shift operators**                                  | Left-to-right            |

| **13**     | `not`                                                                         | **Logical NOT**                                              | Right-to-left            |
| **14**     | `and`                                                                         | **Logical AND**                                              | Left-to-right            |
| **15**     | `or`                                                                          | **Logical OR**                                               | Left-to-right            |
| **16**     | `if – else`                                                                   | **Conditional expressions** (ternary)                        | Right-to-left            |
| **17**     | `lambda`                                                                      | **Lambda expressions**                                       | Right-to-left            |

# Arithmetic
| Operator | Name           | Description                                     | Example  |
| -------- | -------------- | ----------------------------------------------- | -------- |
| `+`      | Addition       | Adds two operands                               | `a + b`  |
| `-`      | Subtraction    | Subtracts second operand from first             | `a - b`  |
| `*`      | Multiplication | Multiplies two operands                         | `a * b`  |
| `/`      | Division       | Divides first operand by second                 | `a / b`  |
| `//`     | Floor Division | Division that results in whole number (floored) | `a // b` |
| `%`      | Modulus        | Returns remainder of division                   | `a % b`  |
| `**`     | Exponentiation | Performs power calculation                      | `a ** b` |

```python
a = 21
b = 10

c = a + b
print("a: {} b: {} a+b: {}".format(a, b, c))

c = a - b
print("a: {} b: {} a-b: {}".format(a, b, c))

c = a * b
print("a: {} b: {} a*b: {}".format(a, b, c))

c = a / b
print("a: {} b: {} a/b: {}".format(a, b, c))

a = 10
b = 5
c = a // b 
print("a: {} b: {} a//b: {}".format(a, b, c))

a = 21
b = 10
c = a % b
print("a: {} b: {} a%b: {}".format(a, b, c))

a = 2
b = 3
c = a ** b 
print("a: {} b: {} a**b: {}".format(a, b, c))
```

# Comparison

| Operator | Name                  | Description                                  | Example  |
| -------- | --------------------- | -------------------------------------------- | -------- |
| `==`     | Equal to              | Returns `True` if both values are equal      | `a == b` |
| `!=`     | Not equal to          | Returns `True` if values are not equal       | `a != b` |
| `>`      | Greater than          | Returns `True` if left is greater than right | `a > b`  |
| `<`      | Less than             | Returns `True` if left is less than right    | `a < b`  |
| `>=`     | Greater than or equal | Returns `True` if left is greater or equal   | `a >= b` |
| `<=`     | Less than or equal    | Returns `True` if left is less or equal      | `a <= b` |

```python
a = 21
b = 10
if (a == b):
   print("Line 1 - a is equal to b")
else:
   print("Line 1 - a is not equal to b")

if (a != b):
   print("Line 2 - a is not equal to b")
else:
   print("Line 2 - a is equal to b")

if (a < b):
   print("Line 3 - a is less than b" )
else:
   print("Line 3 - a is not less than b")

if (a > b):
   print("Line 4 - a is greater than b")
else:
   print("Line 4 - a is not greater than b")

a, b = b, a    #values of a and b swapped. a becomes 10, b becomes 21

if (a <= b):
   print("Line 5 - a is either less than or equal to  b")
else:
   print("Line 5 - a is neither less than nor equal to  b")

if (b >= a):
   print("Line 6 - b is either greater than  or equal to b")
else:
   print("Line 6 - b is neither greater than  nor equal to b")
```

# Assignment

| Operator | Description                    | Example               | Same As      |
| -------- | ------------------------------ | --------------------- | ------------ |
| `=`      | Assign                         | `x = 5`               |              |
| `+=`     | Add and assign                 | `x += 3`              | `x = x + 3`  |
| `-=`     | Subtract and assign            | `x -= 2`              | `x = x - 2`  |
| `*=`     | Multiply and assign            | `x *= 4`              | `x = x * 4`  |
| `/=`     | Divide and assign              | `x /= 2`              | `x = x / 2`  |
| `//=`    | Floor divide and assign        | `x //= 2`             | `x = x // 2` |
| `%=`     | Modulus and assign             | `x %= 3`              | `x = x % 3`  |
| `**=`    | Exponentiate and assign        | `x **= 2`             | `x = x ** 2` |
| `&=`     | Bitwise AND and assign         | `x &= 1`              | `x = x & 1`  |
| `\|=`    | Bitwise OR and assign          | `x \|= 2`             | `x = x \| 2` |
| `^=`     | Bitwise XOR and assign         | `x ^= 3`              | `x = x ^ 3`  |
| `>>=`    | Bitwise right shift and assign | `x >>= 1`             | `x = x >> 1` |
| `<<=`    | Bitwise left shift and assign  | `x <<= 1`             | `x = x << 1` |
| `:=`     | Walrus                         | `print(x := 3)`       | `x = 3`      |

```python
a = 21
b = 10
c = 0
print("a: {} b: {} c : {}".format(a, b, c))
c = a + b
print("a: {}  c = a + b : {}".format(a, c))

c += a
print("a: {} c += a : {}".format(a, c))

c *= a
print("a: {} c *= a : {}".format(a, c))

c /= a 
print("a: {} c /= a : {}".format(a, c))

c = 2
print("a: {} b: {} c : {}".format(a, b, c))
c %= a
print("a: {} c %= a : {}".format(a, c))

c **= a
print("a: {} c **= a : {}".format(a, c))

c //= a
print("a: {} c //= a : {}".format(a, c))
```

# Bitwise

| Operator | Name        | Description                                               | Example (x=6, y=3)                         | Binary Example               |
| -------- | ----------- | --------------------------------------------------------- | ------------------------------------------ | ---------------------------- |
| `&`      | AND         | Sets each bit to 1 if both bits are 1                     | `x & y` → `2`                              | `0110 & 0011 = 0010`         |
| `\|`     | OR          | Sets each bit to 1 if one of the bits is 1                | `x \| y` → `7`                             | `0110 \| 0011 = 0111`        |
| `^`      | XOR         | Sets each bit to 1 if only one of the bits is 1           | `x ^ y` → `5`                              | `0110 ^ 0011 = 0101`         |
| `~`      | NOT         | Inverts all the bits (1's complement)                     | `~x` → `-7`                                | `~0110 = 1001` (in 2’s comp) |
| `<<`     | Left Shift  | Shifts bits to the left by specified number of positions  | `x << 1` → `12`                            | `0110 << 1 = 1100`           |
| `>>`     | Right Shift | Shifts bits to the right by specified number of positions | `x >> 1` → `3`                             | `0110 >> 1 = 0011`           |

```python
a = 20
b = 10

print('a=', a, ':', bin(a), 'b=', b, ':', bin(b))
c = 0

c = a & b
print("result of AND is", c, ':', bin(c))

c = a | b
print("result of OR is", c, ':', bin(c))

c = a ^ b
print("result of EXOR is", c, ':', bin(c))

c = ~a
print("result of COMPLEMENT is", c, ':', bin(c))

c = a << 2
print("result of LEFT SHIFT is", c, ':', bin(c))

c = a >> 2
print("result of RIGHT SHIFT is", c, ':', bin(c))
```

# Logical

| Operator | Description                                          | Example          | Result  |
| -------- | ---------------------------------------------------- | ---------------- | ------- |
| `and`    | Returns `True` if **both** operands are `True`       | `True and False` | `False` |
| `or`     | Returns `True` if **at least one** operand is `True` | `True or False`  | `True`  |
| `not`    | Reverses the Boolean value                           | `not True`       | `False` |

```python
var = 5

print(var > 3 and var < 10)
print(var > 3 or var < 4)
print(not (var > 3 and var < 10))
```

# Membership

| Operator | Description                                    | Example              | Result |
| -------- | ---------------------------------------------- | -------------------- | ------ |
| `in`     | Returns `True` if the value is present         | `'a' in 'apple'`     | `True` |
| `not in` | Returns `True` if the value is **not** present | `'z' not in 'apple'` | `True` |

```python
a = 10
b = 20
list = [1, 2, 3, 4, 5 ]

print("a:", a, "b:", b, "list:", list)

if (a in list):
   print("a is present in the given list")
else:
   print("a is not present in the given list")

if (b not in list):
   print("b is not present in the given list")
else:
   print("b is present in the given list")

c = b / a
print("c:", c, "list:", list)
if (c in list):
   print("c is available in the given list")
else:
   print("c is not available in the given list")
```

# Identity

| Operator | Description                                                          | Example      | Result            |
| -------- | -------------------------------------------------------------------- | ------------ | ----------------- |
| `is`     | Returns `True` if both variables **point to the same object**        | `x is y`     | `True` or `False` |
| `is not` | Returns `True` if both variables **do not point to the same object** | `x is not y` | `True` or `False` |


```python
a = [1, 2, 3, 4, 5]
b = [1, 2, 3, 4, 5]
c = a

print(a is c)
print(a is b)

print(a is not c)
print(a is not b)
```
