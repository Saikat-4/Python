# `'` Single Quote

```python
txt = 'It\'s alright.'
print(txt) 
```

# `"` Double Quote

```python
txt = "We are the so-called \"Vikings\" from the north."
print(txt)
```

# `\` Backslash

```python
txt = "This will insert one \\ (backslash)."
print(txt) 
```

# `\n` New Line

```python
txt = "Hello\nWorld!"
print(txt)
```

# `\t` Tab

```python
txt = "Hello\tWorld!"
print(txt)
```

# `\` NewLine

```python
a = "Hi \
this \
is \
Saikat."
print(a)
```

# `\b` Backspace

```python
txt = "Hello \bWorld!"
print(txt)
```

# `\r` Carriage Return

```python
txt = "Hello\rWorld!"
print(txt)
```

# `\ooo` Octal Value

```python
txt = "\110\145\154\154\157"
print(txt)
```

# `\xhh` Hex Value

```python
txt = "\x48\x65\x6c\x6c\x6f"
print(txt)
```

# `\f` Form Feed

```python
s = "Hello\fWorld"
print(s)
```

# `\a` Bell

```python
s ='Hello\aWorld'
print(s)
```

# `\v` Vertical Tab

```python
s ='Hello\vWorld'
print(s)
```

| Escape Sequence | Meaning                       | Example Output                                |
| --------------- | ----------------------------- | --------------------------------------------- |
| `\\`            | Backslash                     | `\`                                           |
| `\'`            | Single quote                  | `'`                                           |
| `\"`            | Double quote                  | `"`                                           |
| `\a`            | Bell (alert)                  | Triggers a sound (if supported)               |
| `\b`            | Backspace                     | Deletes previous character (in some consoles) |
| `\f`            | Form feed                     | New page (not commonly used)                  |
| `\n`            | Newline                       | Line break                                    |
| `\r`            | Carriage return               | Moves to line start                           |
| `\t`            | Horizontal tab                | Tab space                                     |
| `\v`            | Vertical tab                  | Rarely used vertical space                    |
| `\ooo`          | Octal value                   | Character with octal value                    |
| `\xhh`          | Hex value (2-digit)           | Character with hex value                      |
| `\N{name}`      | Unicode character by name     | e.g., `\N{LATIN SMALL LETTER A}` = `a`        |
| `\uXXXX`        | Unicode 16-bit (4 hex digits) | e.g., `\u00A9` = ©                            |
| `\UXXXXXXXX`    | Unicode 32-bit (8 hex digits) | e.g., `\U0001F600` = 😀                       |
