# Scope in Python
Scope defines where a variable is accessible in the code.

***Types of Scopes (LEGB Rule)***
| Level | Description | Example                                 |
| ----- | ----------- | --------------------------------------- |
| **L** | Local       | Inside the function                     |
| **E** | Enclosing   | Inside nested functions                 |
| **G** | Global      | Defined at the top level                |
| **B** | Built-in    | Python built-in names like `len`, `sum` |

# Nonlocal Keyword
Used to refer to variables in the enclosing (non-global) scope in nested functions.

```python
def outer():
    x = "outer"

    def inner():
        nonlocal x
        x = "inner"
    inner()
    print(x)  # prints 'inner'
```
