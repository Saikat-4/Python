# ***File Opening Modes***

***Primary Modes:***
| Mode  | Description                                                |
| ----- | ---------------------------------------------------------- |
| `'r'` | Read-only (file must exist).                               |
| `'w'` | Write-only (truncates file or creates a new one).          |
| `'a'` | Append-only (creates file if it doesn’t exist).            |
| `'x'` | Create-only (raises `FileExistsError` if the file exists). |

***Modifiers (Can be combined with primary modes):***
| Mode  | Description                               |
| ----- | ----------------------------------------- |
| `'b'` | Binary mode (e.g., images, audio files).  |
| `'t'` | Text mode (default).                      |
| `'+'` | Update mode (both read and write access). |

***Examples of Useful Combinations***
| Mode    | Meaning                   | Use Case                                           |
| ------- | ------------------------- | -------------------------------------------------- |
| `'rt'`  | Read text (default)       | Open a text file for reading.                      |
| `'rb'`  | Read binary               | Read an image or audio file.                       |
| `'wt'`  | Write text                | Overwrite/create a text file.                      |
| `'wb'`  | Write binary              | Write data to a binary file (e.g., save an image). |
| `'a+'`  | Append + Read             | Append and read from a file.                       |
| `'r+'`  | Read + Write              | Read and write, but doesn’t create if missing.     |
| `'w+'`  | Write + Read              | Truncates or creates file for reading and writing. |
| `'rb+'` | Read + Write binary       | Read and modify binary file.                       |
| `'xb'`  | Exclusive binary creation | Create a binary file, error if exists.             |

```python
# Opening a file in read mode
file = open("example.txt", "r")

# Opening a file in write mode
file = open("example.txt", "w")

# Opening a file in append mode
file = open("example.txt", "a")

# Opening a file in binary read mode
file = open("example.txt", "rb")
```

`fileObject.seek(offset, whence)`

***offset*** − This is the position of the read/write pointer within the file.

***whence*** − This is optional and defaults to 0 which means absolute file positioning, other values are 1 which means seek relative to the current position and 2 means seek relative to the file's end.

```python
# Open a file in read-write mode
fo = open("foo.txt", "w+")

# Write initial data to the file
fo.write("This is a rat race")

# Move the read/write pointer to the 10th byte
fo.seek(10, 0)

# Read 3 bytes from the current position
data = fo.read(3)

# Move the read/write pointer back to the 10th byte
fo.seek(10, 0)

# Overwrite the existing content with new text
fo.write('cat')

# Close the file
fo.close()
```

# ***Python - Read Files***

Syntax :--- `file_object.read(size)`

***file_object*** is the file object returned by the open() function.

***size*** is the number of bytes to read from the file. This parameter is optional. If omitted or set to a negative value, the method reads until the end of the file.

```python
# Open the file in read mode
file = open('example.txt', 'r')

# Read the entire content of the file
content = file.read()

# Print the content
print(content)

# Close the file
file.close()
```

Syntax :--- `file_object.readline(size)`

***file_object*** is the file object returned by the open() function.

***size*** is an optional parameter specifying the maximum number of bytes to read from the line. If omitted or set to a negative value, the method reads until the end of the line.

```python
# Open the file in read mode
file = open('example.txt', 'r')

# Read the first line of the file
line = file.readline()

# Print the line
print(line)

# Close the file
file.close()
```

Syntax :--- `file_object.readlines(hint)`

***file_object*** is the file object returned by the open() function.

***hint*** is an optional parameter that specifies the number of bytes to read. If specified, it reads lines up to the specified bytes, not necessarily reading the entire file.

```python
# Open the file in read mode
file = open('example.txt', 'r')

# Read all lines from the file
lines = file.readlines()

# Print the lines
for line in lines:
   print(line, end='')

# Close the file
file.close()
```

```python
# Using the with statement to open a file
with open('example.txt', 'r') as file:
   content = file.read()
   print(content)
```

```python
# Open the file in binary write mode
with open('test.bin', 'wb') as f:
   data = b"Hello World"
   f.write(data)
```

```python
# Open the file in binary read mode
with open('test.bin', 'rb') as f:
   data = f.read()
   print(data.decode(encoding='utf-8'))
```

```python
# Convert the integer to bytes and write to a binary file
n = 25
data = n.to_bytes(8, 'big')

with open('test.bin', 'wb') as f:
   f.write(data)
```

```python
# Read the binary data from the file and convert it back to an integer
with open('test.bin', 'rb') as f:
   data = f.read()
   n = int.from_bytes(data, 'big')
   print(n)
```

```python
import struct

# Define a floating-point number
x = 23.50

# Pack the float into a binary format
data = struct.pack('f', x)

# Open the file in binary write mode and write the packed data
with open('test.bin', 'wb') as f:
   f.write(data)
```

```python
import struct

# Open the file in binary read mode
with open('test.bin', 'rb') as f:
   # Read the binary data from the file
   data = f.read()
    
   # Unpack the binary data to retrieve the float
   x = struct.unpack('f', data)[0]
    
   # Print the float value
   print(x)
```

Syntax :--- `fileObject.seek(offset, whence)`

***offset*** − This is the position of the read/write pointer within the file.

***whence*** − This is optional and defaults to 0 which means absolute file positioning, other values are 1 which means seek relative to the current position and 2 means seek relative to the file's end.

```python
# Open the file in read-write mode
with open("foo.txt", "r+") as fo:
   # Move the read/write pointer to the 10th byte position
   fo.seek(10, 0)
    
   # Read 3 bytes from the current position
   data = fo.read(3)
    
   # Print the read data
   print(data)
```

```python
# Open the file in read-write mode
with open("foo.txt", "r+") as fo:
   # Write data to the file
   fo.write("This is a rat race")

   # Rewind the pointer to the beginning of the file
   fo.seek(0)

   # Read data from the file
   data = fo.read()
   print(data)
```

```python
# Open a file in read-write mode
fo = open("foo.txt", "w+")

# Write initial data to the file
fo.write("This is a rat race")

# Seek to a specific position in the file
fo.seek(10, 0)

# Read a few bytes from the current position
data = fo.read(3)
print("Data read from position 10:", data)

# Seek back to the same position
fo.seek(10, 0)

# Overwrite the earlier contents with new text
fo.write("cat")

# Rewind to the beginning of the file
fo.seek(0, 0)

# Read the entire file content
data = fo.read()
print("Updated file content:", data)

# Close the file
fo.close()
```

**file.writelines(sequence)** :--- Writes a sequence of strings to the file. The sequence can be any iterable object producing strings, typically a list of strings.

Syntax :--- `fileObject.writelines(sequence)`

***sequence*** − This is the Sequence of the strings.

```python
# Open a file in write mode
fo = open("foo.txt", "w")
print("Name of the file: ", fo.name)

# Assuming file has following 5 lines
# This is 1st line
# This is 2nd line
# This is 3rd line
# This is 4th line
# This is 5th line

# Write a new line into the file
seq = ["This is the new line", '\n', "This is another new line"]
line = fo.writelines(seq)

print("Check the file to see the reflected changes")

# Close opened file
fo.close()
```
